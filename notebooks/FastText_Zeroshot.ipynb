```python
import datasets
import pandas as pd
from huggingface_hub import notebook_login

def build_multilingual_benchmark(languages, split="devtest"):
    """
    Downloads and aligns multiple languages from the FLORES+ dataset.
    """
    loaded_datasets = []

    print(f"Loading {len(languages)} languages for the '{split}' split...")

    for lang in languages:
        print(f"  Downloading {lang}...")
        ds = datasets.load_dataset('openlanguagedata/flores_plus', lang, split=split)
        ds = ds.select_columns(['id', 'text']).rename_column('text', lang)
        loaded_datasets.append(ds)

    print("\nMerging languages...")

    merged_ds = loaded_datasets[0]

    for i in range(1, len(loaded_datasets)):
        lang_code = languages[i]
        merged_ds = merged_ds.add_column(lang_code, loaded_datasets[i][lang_code])

    return merged_ds

# Log in to Hugging Face Hub (this will prompt you for your token)
notebook_login()

# Removed 'pli_Latn' (Pali) as it is not supported in FLORES+
target_languages = [
    "eng_Latn", # English (Baseline)
    "sin_Sinh", # Sinhala
    "san_Deva", # Sanskrit
    "tam_Taml", # Tamil
    "hin_Deva", # Hindi
    "ben_Beng", # Bengali
    "arb_Arab", # Arabic (Modern Standard)
    "fra_Latn", # French
    "deu_Latn"  # German
]

# Build the dataset
benchmark_dataset = build_multilingual_benchmark(target_languages, split="devtest")

# View the results as a Pandas DataFrame
df = benchmark_dataset.to_pandas()
print("\nDataset Shape:", df.shape)
print("\nFirst 3 aligned sentences:\n")
print(df.head(3).to_string())
```

```python
import os
import wget
import fasttext
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score, classification_report

# 1. Download the pre-trained fastText LID-176 bin model
model_url = "https://dl.fbaipublicfiles.com/fasttext/supervised-models/lid.176.bin"
model_path = "lid.176.bin"

if not os.path.exists(model_path):
    print("Downloading fastText LID-176 model (~126MB)...")
    wget.download(model_url, model_path)

# 2. Load the model
print("\nLoading fastText model...")
ft_model = fasttext.load_model(model_path)

# 3. Map FLORES+ codes to fastText ISO 639-1 labels
FASTTEXT_LANG_MAP = {
    "eng_Latn": "en",
    "sin_Sinh": "si",
    "san_Deva": "sa",
    "tam_Taml": "ta",
    "hin_Deva": "hi",
    "ben_Beng": "bn",
    "arb_Arab": "ar",
    "fra_Latn": "fr",
    "deu_Latn": "de"
}

# 4. Reshape DataFrame from wide format to long format
df_long_ft = df.melt(
    id_vars=["id"],
    value_vars=list(FASTTEXT_LANG_MAP.keys()),
    var_name="flores_code",
    value_name="text"
)

# Map the true label to match the fastText output format
df_long_ft["true_label"] = df_long_ft["flores_code"].map(FASTTEXT_LANG_MAP)

# fastText strictly expects texts to not have newline characters
# as it treats newlines as end-of-document.
texts = df_long_ft["text"].astype(str).str.replace('\n', ' ').tolist()

# 5. Run Predictions
print(f"Evaluating {len(texts)} samples...")
# ft_model.predict takes a list of strings and returns labels and probabilities
preds, probs = ft_model.predict(texts, k=1)

predictions = []
for p in preds:
    # fastText outputs labels like '__label__en', so we strip the prefix
    pred_label = p[0].replace("__label__", "")
    predictions.append(pred_label)

df_long_ft["predicted_label"] = predictions

# 6. Compute Benchmark Metrics
y_true = df_long_ft["true_label"]
y_pred = df_long_ft["predicted_label"]

# Accuracy across all samples
acc = accuracy_score(y_true, y_pred)

# Macro F1, restricting the evaluation to our 9 target languages
target_labels = list(FASTTEXT_LANG_MAP.values())
macro_f1 = f1_score(y_true, y_pred, average="macro", labels=target_labels)

print("\n" + "="*48)
print("🏆 ZERO-SHOT BENCHMARK RESULTS (fastText LID-176)")
print("="*48)
print(f"Accuracy:  {acc * 100:.2f}%")
print(f"Macro F1:  {macro_f1 * 100:.2f}%")
print("="*48)

# 7. Display Per-Language Classification Report
print("\n📊 Detailed Per-Language Breakdown:\n")
# If fastText predicts a language outside our 9 targets (e.g., 'es' for Spanish),
# restricting the labels array ensures it counts against the target language's recall.
print(classification_report(y_true, y_pred, labels=target_labels, digits=4))
```

```python
import re
import fasttext
import pandas as pd
from huggingface_hub import hf_hub_download
from sklearn.metrics import accuracy_score, f1_score, classification_report

# 1. Download and load the OpenLID-v2 model from Hugging Face
print("Downloading OpenLID-v2 model from Hugging Face...")
openlid_model_path = hf_hub_download(repo_id="laurievb/OpenLID-v2", filename="model.bin")

print("\nLoading OpenLID model into fastText...")
openlid_model = fasttext.load_model(openlid_model_path)

# 2. Define the exact FLORES+ codes we are targeting
# OpenLID-v2 labels natively match the FLORES+ format
target_languages = [
    "eng_Latn",
    "sin_Sinh",
    "san_Deva",
    "tam_Taml",
    "hin_Deva",
    "ben_Beng",
    "arb_Arab",
    "fra_Latn",
    "deu_Latn"
]

# 3. Reshape DataFrame from wide format to long format
df_long_olid = df.melt(
    id_vars=["id"],
    value_vars=target_languages,
    var_name="true_label",
    value_name="text"
)

# 4. Clean text according to OpenLID's recommended preprocessing
# OpenLID performs best when text is lowercased, and non-words/digits are stripped.
def clean_for_openlid(text):
    text = str(text).strip().replace('\n', ' ').lower()
    # Remove digits and punctuation/non-word characters
    text = re.sub(r'[^\w\s]|\d', '', text)
    # Squeeze multiple spaces into a single space
    text = re.sub(r'\s+', ' ', text).strip()
    return text

print("Applying text normalization...")
texts = df_long_olid["text"].apply(clean_for_openlid).tolist()

# 5. Run Predictions
print(f"Evaluating {len(texts)} samples with OpenLID-v2...")
preds, probs = openlid_model.predict(texts, k=1)

predictions = []
for p in preds:
    # OpenLID outputs labels like '__label__eng_Latn', so we strip the prefix
    pred_label = p[0].replace("__label__", "")
    predictions.append(pred_label)

df_long_olid["predicted_label"] = predictions

# 6. Compute Benchmark Metrics
y_true = df_long_olid["true_label"]
y_pred = df_long_olid["predicted_label"]

acc = accuracy_score(y_true, y_pred)
macro_f1 = f1_score(y_true, y_pred, average="macro", labels=target_languages)

print("\n" + "="*48)
print("🏆 ZERO-SHOT BENCHMARK RESULTS (OpenLID-v2)")
print("="*48)
print(f"Accuracy:  {acc * 100:.2f}%")
print(f"Macro F1:  {macro_f1 * 100:.2f}%")
print("="*48)

# 7. Display Per-Language Classification Report
print("\n📊 Detailed Per-Language Breakdown:\n")
# Using labels=target_languages ensures strict evaluation against your expected classes
print(classification_report(y_true, y_pred, labels=target_languages, digits=4))
```
