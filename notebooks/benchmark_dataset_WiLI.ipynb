### WiLI-2018 Benchmark Dataset

Load the WiLI-2018 dataset and inspect the Sinhala (`sin`) examples.

```python
import pandas as pd

DATA_DIR = "../datasets/wili-2018"

#Tab issue-read the whole line as one example
with open(f"{DATA_DIR}/x_train.txt", encoding="utf-8") as f:
    x_train_lines = f.read().splitlines()

with open(f"{DATA_DIR}/y_train.txt", encoding="utf-8") as f:
    y_train_lines = f.read().splitlines()

assert len(x_train_lines) == len(y_train_lines), "x_train/y_train line count mismatch"

train_df = pd.DataFrame({"text": x_train_lines, "label": y_train_lines})
train_df.shape
```

```python
import linecache
import random


def spot_check(line_no: int) -> None:
    """Cross check a 1-indexed line directly against the raw files"""
    raw_text =linecache.getline(f"{DATA_DIR}/x_train.txt", line_no).rstrip("\n")
    raw_label=linecache.getline(f"{DATA_DIR}/y_train.txt", line_no).rstrip("\n")
    df_text = train_df.loc[line_no - 1, "text"]
    df_label = train_df.loc[line_no - 1, "label"]
    assert raw_text == df_text, f"text mismatch at line {line_no}"
    assert raw_label == df_label, f"label mismatch at line {line_no}"


#suspicious row I found when manual checking(line29245 in X_train.csv)
for line_no in [29245, *random.sample(range(1, len(train_df) + 1), 5)]:
    spot_check(line_no)

print("All spot checks passed: DataFrame rows align exactly with the raw file lines.")
```

```python
sinhala_df = train_df[train_df["label"] == "sin"].reset_index(drop=True)

print(f"Number of Sinhala examples: {len(sinhala_df)}")
sinhala_df.head(10)

sinhala_df.to_csv(f"{DATA_DIR}/sinhala_examples.csv", index=False)
```
