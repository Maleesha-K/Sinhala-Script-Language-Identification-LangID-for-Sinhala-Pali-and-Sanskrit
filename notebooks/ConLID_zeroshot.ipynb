```python
import datasets
import pandas as pd
from huggingface_hub import notebook_login

def build_multilingual_benchmark(languages, split="devtest"):
    """
    Downloads and aligns multiple languages from the FLORES+ dataset.
    """
    loaded_datasets = []

    print(f"Loading {len(languages)} languages for the '{split}' split...")

    for lang in languages:
        print(f"  Downloading {lang}...")
        ds = datasets.load_dataset('openlanguagedata/flores_plus', lang, split=split)
        ds = ds.select_columns(['id', 'text']).rename_column('text', lang)
        loaded_datasets.append(ds)

    print("\nMerging languages...")

    merged_ds = loaded_datasets[0]

    for i in range(1, len(loaded_datasets)):
        lang_code = languages[i]
        merged_ds = merged_ds.add_column(lang_code, loaded_datasets[i][lang_code])

    return merged_ds

# Log in to Hugging Face Hub (this will prompt you for your token)
notebook_login()

# Removed 'pli_Latn' (Pali) as it is not supported in FLORES+
target_languages = [
    "eng_Latn", # English (Baseline)
    "sin_Sinh", # Sinhala
    "san_Deva", # Sanskrit
    "tam_Taml", # Tamil
    "hin_Deva", # Hindi
    "ben_Beng", # Bengali
    "arb_Arab", # Arabic (Modern Standard)
    "fra_Latn", # French
    "deu_Latn"  # German
]

# Build the dataset
benchmark_dataset = build_multilingual_benchmark(target_languages, split="devtest")

# View the results as a Pandas DataFrame
df = benchmark_dataset.to_pandas()
print("\nDataset Shape:", df.shape)
print("\nFirst 3 aligned sentences:\n")
print(df.head(3).to_string())
```

```python
import os
import sys
import pandas as pd
from huggingface_hub import snapshot_download
from sklearn.metrics import accuracy_score, f1_score, classification_report
from tqdm.auto import tqdm

# 1. Ensure Colab can find the 'model.py' wrapper we just cloned
if not os.path.exists('model.py'):
    os.chdir('/content/language-identification')
sys.path.append(os.getcwd())

from model import ConLID

# 2. Download the ConLID model checkpoints from Hugging Face
print("Downloading ConLID model...")
checkpoint_dir = "checkpoints/conlid"
snapshot_download(repo_id="epfl-nlp/ConLID", local_dir=checkpoint_dir)

# 3. Load the Model
print("\nLoading ConLID model...")
conlid_model = ConLID.from_pretrained(dir=checkpoint_dir)

# 4. Define the exact FLORES+ codes we are targeting
target_languages = [
    "eng_Latn",
    "sin_Sinh",
    "san_Deva",
    "tam_Taml",
    "hin_Deva",
    "ben_Beng",
    "arb_Arab",
    "fra_Latn",
    "deu_Latn"
]

# 5. Reshape your existing DataFrame (assuming 'df' is still in memory)
df_long_conlid = df.melt(
    id_vars=["id"],
    value_vars=target_languages,
    var_name="true_label",
    value_name="text"
)

# Convert to a list of strings
texts = df_long_conlid["text"].astype(str).tolist()

# 6. Run Predictions
print(f"Evaluating {len(texts)} samples with ConLID...")
predictions = []

# ConLID's wrapper predicts one by one.
# We use tqdm to display a progress bar since we are iterating.
for text in tqdm(texts):
    # predict() returns a tuple like: (['eng_Latn'], [0.9709...])
    pred_result = conlid_model.predict(text, k=1)

    # Extract the top label
    top_label = pred_result[0][0]
    predictions.append(top_label)

df_long_conlid["predicted_label"] = predictions

# 7. Compute Benchmark Metrics
y_true = df_long_conlid["true_label"]
y_pred = df_long_conlid["predicted_label"]

acc = accuracy_score(y_true, y_pred)
macro_f1 = f1_score(y_true, y_pred, average="macro", labels=target_languages)

print("\n" + "="*48)
print("🏆 ZERO-SHOT BENCHMARK RESULTS (ConLID)")
print("="*48)
print(f"Accuracy:  {acc * 100:.2f}%")
print(f"Macro F1:  {macro_f1 * 100:.2f}%")
print("="*48)

# 8. Display Per-Language Classification Report
print("\n📊 Detailed Per-Language Breakdown:\n")
# Using labels=target_languages ensures strict evaluation against your expected classes
print(classification_report(y_true, y_pred, labels=target_languages, digits=4))
```
