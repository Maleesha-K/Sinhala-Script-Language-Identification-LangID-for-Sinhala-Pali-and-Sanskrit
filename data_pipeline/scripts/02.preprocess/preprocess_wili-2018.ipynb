```python
input_dir = 'datasets/raw_download/wili-2018'
output_file = 'datasets/preprocessed/wili-2018.jsonl'
```

```python
import os
# Auto-resolve the project root if running manually
if not os.path.exists("Makefile") and os.path.exists("../../Makefile"):
    os.chdir("../../")

import os
import json

print(f"Preprocessing data from {input_dir}...")

splits = [("x_train.txt", "y_train.txt"), ("x_test.txt", "y_test.txt")]

missing = [
    fname for pair in splits for fname in pair
    if not os.path.exists(os.path.join(input_dir, fname))
]
if missing:
    print(f"Error: missing files in {input_dir}: {missing}")
else:
    os.makedirs(os.path.dirname(output_file), exist_ok=True)
    processed_count = 0
    with open(output_file, "w", encoding="utf-8") as f_out:
        for x_name, y_name in splits:
            x_path = os.path.join(input_dir, x_name)
            y_path = os.path.join(input_dir, y_name)

            with open(x_path, encoding="utf-8") as f_x, open(y_path, encoding="utf-8") as f_y:
                texts = f_x.read().splitlines()
                labels = f_y.read().splitlines()

            assert len(texts) == len(labels), f"{x_name}/{y_name} line count mismatch"

            for text, label in zip(texts, labels):
                standard_record = {
                    "text": text,
                    "label": label,
                    "source": "wili-2018",
                }
                f_out.write(json.dumps(standard_record, ensure_ascii=False) + "\n")
                processed_count += 1

    print(f"Successfully processed {processed_count} records to {output_file}")
```
